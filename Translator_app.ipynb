```python
import sys

# ==========================================
# 1. DIAGNOSTICS & SETUP
# ==========================================

# Ensure deep_translator is installed when this cell runs
!pip install deep-translator

# Removed the deep_translator import error check since it's installed and not needed.
from deep_translator import GoogleTranslator

try:
    import gradio as gr
except ImportError:
    print("\n❌ ERROR: 'gradio' is not installed.")
    print("👉 Fix it by running this in your terminal: pip install gradio\n")
    sys.exit()

# ==========================================
# 2. TRANSLATION LOGIC
# ==========================================
# Updated with standard ISO 639-1 language codes
SUPPORTED_LANGUAGES = {
    "Telugu": "te",
    "Tamil": "ta",
    "Malayalam": "ml",
    "Hindi": "hi",
    "Japanese": "ja",
    "English": "en"
}

def translate_text(text, target_language_name):
    """Takes input text and language name, returns translated text."""
    if not text.strip():
        return "Please enter some text to translate."

    try:
        # Fetch the correct two-letter code for the backend
        target_code = SUPPORTED_LANGUAGES[target_language_name]

        # Initialize and translate
        translator = GoogleTranslator(source='auto', target=target_code)
        translated = translator.translate(text)
        return translated
    except Exception as e:
        return f"An error occurred during translation: {e}"

# ==========================================
# 3. WEB APP INTERFACE (GRADIO)
# ==========================================
print("🚀 Starting the Translator Web App...")

app = gr.Interface(
    fn=translate_text,
    inputs=[
        gr.Textbox(lines=5, placeholder="Type text here...", label="Source Text"),
        gr.Dropdown(choices=list(SUPPORTED_LANGUAGES.keys()), value="Hindi", label="Target Language")
    ],
    outputs=gr.Textbox(lines=5, label="Translated Text"),
    title="🌍 AI Translator App",
    description="Type any text, choose a language, and the AI will translate it instantly."
)

if __name__ == "__main__":
    app.launch(debug=True)
```
