```python
from spellchecker import SpellChecker
import re

class AutoCorrectTool:
    def __init__(self):
        self.spell = SpellChecker()

    def correct_text(self, text):
        corrections = []

        # Split text while preserving spaces and punctuation
        tokens = re.findall(r"\w+|[^\w\s]|\s+", text)

        result = []

        for token in tokens:
            if token.isalpha():

                # Keep original capitalization
                original = token
                lower_word = token.lower()

                if lower_word not in self.spell:
                    corrected = self.spell.correction(lower_word)

                    if corrected:
                        corrections.append(
                            {
                                "original": original,
                                "corrected": corrected
                            }
                        )

                        if original[0].isupper():
                            corrected = corrected.capitalize()

                        result.append(corrected)
                    else:
                        result.append(original)
                else:
                    result.append(original)

            else:
                result.append(token)

        return "".join(result), corrections


def main():
    tool = AutoCorrectTool()

    print("=" * 50)
    print("AUTO CORRECT TOOL")
    print("=" * 50)

    while True:
        text = input("\nEnter text (or 'quit'): ")

        if text.lower() == "quit":
            break

        corrected_text, changes = tool.correct_text(text)

        print("\nCorrected Text:")
        print(corrected_text)

        print("\nCorrections:")
        if changes:
            for c in changes:
                print(
                    f"{c['original']}  -->  {c['corrected']}"
                )
        else:
            print("No spelling errors found.")


if __name__ == "__main__":
    main()
```
